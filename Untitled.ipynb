```python
import pandas as pd

allyears = []

for year in range(1880,2019):
  path = f'names/yob{year}.txt'
  df = pd.read_csv(path, names=['name','sex','number'])
  df['year'] = year
  allyears.append(df)

allyears = pd.concat(allyears, ignore_index=True)
allyears.info()
```

```python
import matplotlib.pyplot as plt

df = pd.read_csv('allyears.csv.gz')

allyears_indexed = df.set_index(['sex','name','year']).sort_index()

mary_f = allyears_indexed.loc[('F','Mary')]
mary_f.plot()
plt.show()
```

```python
total_births = df.groupby('year')['number'].sum()
mary_norm = mary_f['number'] / total_births
mary_norm.plot()
plt.show()
```

```python
def plotname(sex, name):
    data = allyears_indexed.loc[(sex, name)]
    plt.plot(data.index, data['number'], label=name)
    plt.title(f"{sex}/{name} popularity over time")
    plt.legend()
    plt.show()
```

```python
def comparenames(sex, names):
    for name in names:
        data = allyears_indexed.loc[(sex, name)]
        plt.plot(data.index, data['number'], label=name)
    plt.title(f"{sex} name comparison")
    plt.legend()
    plt.show()
```
